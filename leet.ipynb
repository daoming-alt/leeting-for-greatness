```python
# problem 2. Add Two Numbers
type IntList = list[int]
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    def print_list(self):
        while self is not None:
            print('[{}]->'.format(self.val), end=' ')
            self=self.next
            if self is None:
                print('[#]')
                break

def create_list(l:IntList) -> ListNode:
    """returns start of list
    """
    start=ListNode(l[0])
    pointer = start
    left = l[1:]
    for i in range(len(left)):
        pointer.next = ListNode(left[i])
        pointer = pointer.next

    return start

def read_list_value_to_str(l:ListNode) -> str:
    if l is None:
        return ""
    else:
        s=""
        while True:
            s = s + str(l.val)
            l=l.next
            if l is None:
                break

        return s
        
```

```python
l,l2 = create_list([2,4,3]), create_list([5,6,4])
l.print_list()
l2.print_list()
lVal, l2Val = read_list_value_to_str(l), read_list_value_to_str(l2)
lVal_reversed, l2Val_reversed = lVal[::-1], l2Val[::-1]
ret_val = str(eval(lVal_reversed)+eval(l2Val_reversed))[::-1]
ans=create_list([eval(r) for r in ret_val])
print('######################')
ans.print_list()
type(ans)
### accepted on 3:10/25 correction initial commit
```

```python
# 5. Longest Palindromic Substring
def is_palindromic(s):
    return s == s[::-1]


s="babad"
start,end = 0, len(s)
currMax=(-1,-1)
while start < len(s) and end >= 0:
    current=s[start:end]
    #print(current,is_palindromic(current),sep=' = ', end='\n')
    if is_palindromic(current) and len(current) > currMax[1]-currMax[0]:
        currMax=(start, end)
    start=start+1
    end = end-1
print(currMax)
s[currMax[0]:currMax[1]]
```
